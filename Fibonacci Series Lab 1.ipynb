```python
import numpy as np

def get_fibonacci_series(n):
    """
    Generates the first n Fibonacci numbers using NumPy arrays.
    """
    if n <= 0:
        return np.array([])
    elif n == 1:
        return np.array([0])
    
    # 1. Pre-allocate a NumPy array of zeros
    # We use dtype='object' because Fibonacci numbers grow HUGE very fast.
    # Standard int64 will overflow around the 93rd Fibonacci number.
    fib_series = np.zeros(n, dtype='object')
    
    # 2. Set the starting values
    fib_series[0] = 0
    fib_series[1] = 1
    
    # 3. Calculate the series
    # We still need a loop because n depends on (n-1) and (n-2)
    for i in range(2, n):
        fib_series[i] = fib_series[i-1] + fib_series[i-2]
        
    return fib_series

def get_nth_fibonacci_matrix(n):
    """
    BONUS: Uses NumPy Matrix math to find the Nth number 
    WITHOUT calculating the whole series.
    Math: [[1,1],[1,0]] ^ (n-1)
    """
    if n == 0: return 0
    if n == 1: return 1
    
    # Define the Fibonacci Matrix
    matrix = np.matrix([[1, 1], [1, 0]], dtype='object')
    
    # Calculate Matrix to the power of (n-1)
    # NumPy handles this matrix multiplication incredibly fast
    result_matrix = np.linalg.matrix_power(matrix, n - 1)
    
    # The top-left number is the Nth Fibonacci number
    return result_matrix[0, 0]

# --- Driver Code ---
if __name__ == "__main__":
    print("NumPy Fibonacci Generator ")
    
    try:
        count = int(input("How many Fibonacci numbers do you want? "))
        
        # Get the Series
        result = get_fibonacci_series(count)
        print(f"\nSeries (First {count} numbers):")
        print(result)
        
        # Demonstrate the Matrix Trick (Great for single large numbers)
        print(f"\n Matrix Calculation Check ")
        nth_val = get_nth_fibonacci_matrix(count-1) # Check the last number
        print(f"The last number (calculated via Matrix Power) is: {nth_val}")
        
    except ValueError:
        print("Please enter a valid integer.")
```
